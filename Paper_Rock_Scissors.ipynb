```python
wins_against = {"paper":"rock",
                "rock":"scissors",
                "scissors":"paper"}
player1 = ""
player2 = ""

score1 = 0
score2 = 0

def greeting():
  print("Welcome to the game!")
  print("Please choose out of the following options : (rock,paper,scissors)")


def conclusion(score1, score2):
  print("Game Over")
  if score1 > score2:
    print("Player number 1 won!")
  elif score1 < score2:
    print("Player number 2 won!")


while True:
  greeting()

  player1 = input("Player number one choice :")
  player2 = input("Player number two choice :")
  if player1 not in ("rock","paper","scissors") or player2 not in ("rock","paper","scissors"):
    print("Incorrect Input! Try again.")
    continue
  wrong_choice = wins_against[player1]

  if player2 == wrong_choice:
    score1+=1
    print("Player 1 won this round!")
  elif player1 != player2:
    score2 += 1
    print("Player 2 won this round!")
  elif player1 == player2:
    print("Tie!")


  print("Do you want to continue? Please enter yes or no")
  isContinue = input("Enter:")
  if isContinue == "no":
    break

conclusion(score1, score2)



```
